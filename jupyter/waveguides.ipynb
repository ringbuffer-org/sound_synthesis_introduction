# Ideal String with two Delay Lines


Based on the previously introduced discrete solution for the wave equation,
an ideal, lossless string can be implemented using two delay Lines with direct coupling.
The left-traveling and right-traveling wave are connected end-to-end.
Output samples of each delay Line are direcly inserted to the input of the counterpart with an additional inversion of the sign:

```python
###################################################################
# function for plucking the string

def pluck(L,P):
    
    x_L = np.zeros(L);
    x_R = np.zeros(L);
    
    x_L[:P] = np.linspace(0,1,P)
    x_R[:P] = np.linspace(0,1,P)
    
    x_L[P:] = np.linspace(1,0,L-P)
    x_R[P:] = np.linspace(1,0,L-P)
    
    return x_L, x_R
```

# The Excitation Function

The waveguides can be initiated with any arbitrary excitation function. 
This may vary, depending on the excitation principle of the instrument to be modeled.
For a plucked string, the excitation can be a triangular function with a maximum at the plucking point $p$.
Both waveguides (left/right travelling) are initiated with the same function:

$$
y[i] = 
\begin{cases}
\frac{i}{p}         & \text{for } i \leq p \\
1-\frac{i-p}{N-p}   & \text{for } i > p \\
\end{cases}
$$

# Oscillation


When both waveguides are shifted by one sample each $\frac{1}{f_s}$ seconds, the ideal string is oscillating
with a frequency of $f_0 = \frac{f_s}{N}$. It will oscillate continuously. In fact it is a superposition of two oscillators with the waveform defined by the excitation function.

```python


###################################################################

fs        = 48000

# length of the delay lines:
L         = 50

# feedback gain:
g         = 0.95

# pluck position:
pluck_pos = 10

# pickup position:
pick      = 5

# filter length:
N         = 20

pick_pos  = 10

n_out     = 300

x_L, x_R  = pluck(L,10)


y    = np.array([])

# the filter is a simple moving average
filt = np.zeros(N)


###################################################################

fig    = plt.figure()
fig.tight_layout(pad=3.0)

ax     = fig.add_subplot(2, 1, 1)
line,  = ax.plot(x_L,color=[0.5,0.3,0.3]);
#ax.set_title('Left-travelling Delay Line')
ax.set_xlabel("$i$");
ax.set_ylabel("$y^+$");
ax.set_ylim([-1.2, 1.2])


ax2    = fig.add_subplot(2, 1, 2)
rplot, = ax2.plot(x_R,color=[0.3,0.5,0.3]);
#ax2.set_title('Right-travelling Delay Line')
ax2.set_xlabel("$i$");
ax2.set_ylabel("$y^-$");
ax2.set_ylim([-1.2, 1.2])


###################################################################
# function: - get the next output sample
#           - shift all buffers

def next_step(x_L, x_R, filt, g, pick):
    
    # delay line outputs
    l_out = x_L[0]
    r_out = x_R[len(x_R)-1]    
    
    
    # shift all arrays
    x_L   = np.roll(x_L,-1)
    x_R   = np.roll(x_R,1)    
   
    
    # insert output values
    x_L[len(x_L)-1] = -r_out
    x_R[0]          = -l_out
    
    out =  x_L[pick] + x_R[pick]
    
    return x_L, x_R, filt, out


###################################################################
 

def animate(frame):
       
    global x_L
    global x_R
    global y
    global filt
    global line
    global L
    global pick_pos
    global g
    global fig 
    
    # print(y)
    # print(x_L[0:5])

    # for idx in(i):
    line.set_data(np.linspace(0, L, L), x_L)
    rplot.set_data(np.linspace(0, L, L), x_R)
    
    x_L, x_R, filt, out = next_step(x_L, x_R, filt, g, pick_pos)
    y = np.append(y,out)


 
    #print(len(x_L))
    return (line,)    


###################################################################

def init():
    
    global line
    
    line.set_data([], [])
    return (line,)


anim = animation.FuncAnimation(fig, animate, init_func=init,
                               frames=2*L, interval=50)

plt.close();

HTML(anim.to_jshtml())
#HTML(anim.to_html5_video());

```
